# Optional: Generate summary visualisations for the stakeholder presentation

```python
import matplotlib.pyplot as plt

# Bar chart comparing model performance
models = ['Logistic Reg', 'Gradient Boosting', 'Random Forest', 'LightGBM (Tuned)']
scores = [0.9140, 0.9893, 0.9899, 0.9926]

plt.figure(figsize=(10, 5))
plt.bar(models, scores, color=['#4C72B0', '#55A868', '#C44E52', '#8172B3'], edgecolor='black')
plt.title('Model Performance Comparison (ROC-AUC)', fontsize=14)
plt.ylabel('ROC-AUC Score')
plt.ylim(0.85, 1.05)

# Add target line
plt.axhline(y=0.90, color='r', linestyle='--', label='Business Target (0.90)')
plt.legend()

for i, v in enumerate(scores):
    plt.text(i, v + 0.005, f'{v:.4f}', ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
```
